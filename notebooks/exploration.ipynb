```python
%pip install pandas
import pandas as pd

cond = pd.read_csv("../data/raw/conditions.csv")

# All diabetes-related codes present, with counts
diab = cond[cond["DESCRIPTION"].str.contains("diabet", case=False, na=False)]
print(diab.groupby(["CODE", "DESCRIPTION"]).size().sort_values(ascending=False))

# Number of unique patients with Type 2 diabetes
t2d = cond[cond["CODE"] == 44054006]
print("T2D patients:", t2d["PATIENT"].nunique())
```
