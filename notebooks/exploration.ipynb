```python
%pip install pandas
import pandas as pd

cond = pd.read_csv("../data/raw/conditions.csv")

# All diabetes-related codes present, with counts
diab = cond[cond["DESCRIPTION"].str.contains("diabet", case=False, na=False)]
print(diab.groupby(["CODE", "DESCRIPTION"]).size().sort_values(ascending=False))

# Number of unique patients with Type 2 diabetes
t2d = cond[cond["CODE"] == 44054006]
print("T2D patients:", t2d["PATIENT"].nunique())
```

```python
# The 93 people with Type 2 diabetes
t2d_ids = t2d["PATIENT"].unique()

# Only their diagnoses
t2d_cond = cond[cond["PATIENT"].isin(t2d_ids)]

# The complications we care about
complications = {
    127013003: "Kidney disease due to diabetes",
    90781000119102: "Microalbuminuria (a little protein in urine)",
    157141000119108: "Proteinuria (more protein in urine)",
    368581000119106: "Nerve damage",
    1551000119108: "Eye damage (early)",
    97331000119101: "Eye damage (macular edema)",
    1501000119109: "Eye damage (advanced)",
}

# For each complication, count how many of the 93 patients have it
for code, name in complications.items():
    n = t2d_cond[t2d_cond["CODE"] == code]["PATIENT"].nunique()
    print(f"{name}: {n} of {len(t2d_ids)} patients")
```
